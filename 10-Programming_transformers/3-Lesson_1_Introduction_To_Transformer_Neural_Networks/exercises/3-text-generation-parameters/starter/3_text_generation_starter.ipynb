```python
# You can use this prompt or try something else!
prompt = "During the latest presentation OpenAI"
# A good model for this exercise, but feel free to use another model: https://huggingface.co/models?pipeline_tag=text-generation&sort=downloads
model = "openai-community/gpt2-large"
```

```python
# Step 1 - Set Up the Text Generation Pipeline

# Import "pipeline" function
from transformers import pipeline

# Create a pipeline for text generation, selected model, and "device=0" (to use CUDA)
generator = pipeline("text-generation", model=model, device=0)
```

```python
# Step 2 - Generate text with the default settings

# Generate text with the selected prompt
# Recommended parameters:
# max_length=100 - generate up to 100 tokens
# num_return_sequences=1 - only return one generated sequence
generated_text = generator(prompt, max_length=100, num_return_sequences=1)
print(generated_text)
```

```python
# Step 3 - Experiment with Different Parameters

# Now try different parameters

# Try text generation with "do_sample" parameter equal to `True` or `False`
generated_text2 = generator(prompt, max_length=200, num_return_sequences=2, do_sample=True)
print(generated_text)
```

```python
# Try text generation using "Beam-search strategy"
# Parameters:
# do_sample=False
# num_beams - try different values
generated_text3 = generator(prompt, max_length=200, num_return_sequences=1, do_sample=False, num_beams=3)
print(generated_text3)
```

```python
# Try text generation using "Beam-search multinomial sampling"
# Parameters:
# do_sample=True
# num_beams - try different values
generated_text4 = generator(prompt, max_length=200, num_return_sequences=1, do_sample=True, num_beams=2)
print(generated_text4)
```

```python
# Try text generation with different "top_k" values. E.g. from 1 to 500
generated_text5 = generator(prompt, max_length=200, num_return_sequences=1, do_sample=True, top_k=5)
print("top_k=5:", generated_text5)

generated_text6 = generator(prompt, max_length=200, num_return_sequences=1, do_sample=True, top_k=50)
print("top_k=50:", generated_text6)

generated_text7 = generator(prompt, max_length=200, num_return_sequences=1, do_sample=True, top_k=500)
print("top_k=500:", generated_text7)
```

```python
# Try text generation with different "temperature" values. E.g. from 0.1 to 3.0
generated_text8 = generator(prompt, max_length=200, num_return_sequences=1, do_sample=True, temperature=0.1)
print("temperature=0.1:", generated_text8)

generated_text9 = generator(prompt, max_length=200, num_return_sequences=1, do_sample=True, temperature=1.0)
print("temperature=1:", generated_text9)

generated_text10 = generator(prompt, max_length=200, num_return_sequences=1, do_sample=True, temperature=2.0)
print("temperature=2:", generated_text10)

generated_text11 = generator(prompt, max_length=200, num_return_sequences=1, do_sample=True, temperature=3.0)
print("temperature=3:", generated_text11)
```
